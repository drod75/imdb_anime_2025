# Data Transformation

This notebook focuses on reading in the data from the IMDB Database, and filtering it for specifically data that is related to anime releases in 2026.

```python
import pandas as pd
```

```python
ORIGIN = "https://datasets.imdbws.com/"
akas = "title.akas.tsv.gz"
titles = "title.basics.tsv.gz"
crew = "title.crew.tsv.gz"
episodes = "title.episode.tsv.gz"
principals = "title.principals.tsv.gz"
ratings = "title.ratings.tsv.gz"
names = "name.basics.tsv.gz"
```

```python
tsv_config = {
    "sep": "\t",
    "na_values": "\\N",
    "compression": "gzip",
    "low_memory": False,
}

akas_df = pd.read_csv(f"{ORIGIN}{akas}", **tsv_config)
titles_df = pd.read_csv(f"{ORIGIN}{titles}", **tsv_config)
crew_df = pd.read_csv(f"{ORIGIN}{crew}", **tsv_config)
episodes_df = pd.read_csv(f"{ORIGIN}{episodes}", **tsv_config)
principals_df = pd.read_csv(f"{ORIGIN}{principals}", **tsv_config)
ratings_df = pd.read_csv(f"{ORIGIN}{ratings}", **tsv_config)
names_df = pd.read_csv(f"{ORIGIN}{names}", **tsv_config)
```
